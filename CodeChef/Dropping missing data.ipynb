# Dropping missing data

Problem : https://www.codechef.com/learn/course/pandas/LPDP06/problems/LPDPR28

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'A': [1, 2, np.nan, 4, 5],
    'B': [5, np.nan, np.nan, None, 10],
    'C': ['a', 'b', 'c', None, 'e']
})
                   
print(df)  

# Drop rows where column A has missing values

dropa=df.dropna(subset="A")
print(dropa)


# Drop rows with less than 2 non-null values

less_then_2_num=dropa.dropna(subset='C')
print(less_then_2_num)

# Drop rows where both Column A and Column B has missing values
print(dropa)


```
