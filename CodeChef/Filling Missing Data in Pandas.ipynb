```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
    'A': [1, 2, np.nan, 4, 5],
    'B': [5, np.nan, np.nan, None, 10],
    'C': ['a', 'b', 'c', None, 'e']
})
                   
print(df)  


df_filled = df.fillna({'A': 0, 'B': 99, 'C': 'Unknown'})
print(df_filled)


df_bfilled = df.fillna({'B':5})
print(df_bfilled)
new=df.copy()
new=new.fillna({'A': (new['A'].median()),'B':(new['B'].mean()), 'C':'Unknown'})
print(new)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'A': [1, 2, np.nan, 4, 5],
    'B': [5, np.nan, np.nan, None, 10],
    'C': ['a', 'b', 'c', None, 'e']
})
print(df)  

df_filled = df.fillna(0)
df_filled = df.fillna({'A': 0, 'B': 99, 'C': 'Unknown'})
print(df_filled)
df_ffilled = df.ffill()

df_bfilled = df.bfill()
df['A'].fillna(df['A'].mean(), inplace=True)
df['B'].fillna(df['B'].median(), inplace=True)
```
