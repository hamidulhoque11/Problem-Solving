# GroupBy Operation

problem : https://www.codechef.com/learn/course/pandas/LPD10/problems/LPDPR46

```python
import pandas as pd
import numpy as np

# Create a sample DataFrame
df = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Value1': [10, 20, 30, 40, 50, 60],
    'Value2': [100, 200, 300, 400, 500, 600]
})

# Basic groupby and aggregation
#mean
print("Mean of Value1 for each Category:")
print(df.groupby('Category')['Value1'].mean())

print("Mean of Values2 for each Category:")
print(df.groupby('Category')['Value2'].mean())
#sum
print("Sum of Value1 for each Category:")
print(df.groupby('Category')['Value1'].sum())

print("Sum of Value2 for each Category:")
print(df.groupby('Category')['Value2'].sum())

#Size
print("Size of Value1 for each Category:")
print(df.groupby('Category')['Value1'].size())

print("Size of Value2 for each Category:")
print(df.groupby('Category')['Value2'].size())

#Count
print("Count of Value1 for each Category:")
print(df.groupby('Category')['Value1'].count())

print("Count of Value2 for each Category:")
count_v2=df.groupby('Category')
print(count_v2['Value2'].count())



#min
print("Min of Value1 for each Category:")
max_v1=df.groupby('Category')
print(max_v1['Value1'].min())

print("Min of Value2 for each Category:")
min_v2=df.groupby('Category')
print(count_v2['Value2'].min())
#max

print("Max of Value1 for each Category:")
print(df.groupby('Category')['Value1'].max())

print("Max of Value2 for each Category:")
print(df.groupby('Category')['Value2'].max())

#aggregation 
print("aggregation Function of Value1 for each Category:")
print(df.groupby('Category')['Value1'].agg(['sum', 'mean', 'max', 'min']))

print("aggregation Funciton of Value2 for each Category:")
print(df.groupby('Category')['Value2'].agg(['sum', 'mean', 'max', 'min']))






```
