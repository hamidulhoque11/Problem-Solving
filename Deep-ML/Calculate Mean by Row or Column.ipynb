```python
def calculate_matrix_mean(matrix: list[list[float]], mode: str) -> list[float]:
	mat=np.array(matrix)
	count=0
	if mode=='column':
		count=0
	else :
		count=1
	means=mat.mean(axis=count).round(1).astype(float)
	return means

print(calculate_matrix_mean([[1, 2, 3], [4, 5, 6], [7, 8, 9]], 'column'))
```
