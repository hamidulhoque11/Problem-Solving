```python
import numpy as np
def matrix_dot_vector(a: list[list[int|float]], b: list[int|float]) -> list[int|float]:
	a=np.array(a)
	b=np.array(b)
	if b.shape[0] == a.shape[1]:
		dot = np.dot(a,b)
		return dot
	else:
     		return -1
 
print(matrix_dot_vector(a = [[1, 2], [2, 4]], b = [1, 2]))
```
