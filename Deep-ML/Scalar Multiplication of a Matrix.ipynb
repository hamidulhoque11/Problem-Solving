```python
import numpy as np
def scalar_multiply(matrix: list[list[int|float]], scalar: int|float) -> list[list[int|float]]:
	mat=np.array(matrix)
	val=np.array(scalar)
	mul=mat*val
	return mul
print(scalar_multiply(matrix = [[1, 2], [3, 4]], scalar = 2))
```
