```python
string=input().split()
count_a=0
count_b=0
total_count=0

for x in string:
    x = x.upper()
    if x=="A":
        total_count += 1
        count_a+=1
    elif x=="B":
        count_b+=1
        total_count+=1

predictions_a = (count_a/total_count)*100
predictions_b = (count_b/total_count)*100
if predictions_a > 70 or predictions_b > 70  :
    print("Biased Model")
else:
    print("Fair Model")
```
