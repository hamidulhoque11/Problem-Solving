```python
brightness = list(map(int,(input().split())))
sum=0
for x in brightness:
    sum+=x
avg_brightness=sum/len(brightness)
if avg_brightness < 85:
    print("Dark Image")
elif avg_brightness >= 85 and avg_brightness < 170:
    print("Normal Image")
elif avg_brightness > 170:
    print("Bright Image")
```
