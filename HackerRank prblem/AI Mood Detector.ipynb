```python
string=input().split()
happy_count=0
sad_count=0
for x in string:
    x.lower()
    if x=="happy" or x =="joy" or x=="smile":
        happy_count+=1
    elif x=="sad" or x =="cry" or x=="angry":
            sad_count+=1
if happy_count>sad_count:
    print("Happy Mood")
elif sad_count>happy_count:
    print("Sad Mood")
else:
    print("Neutral Mood")
```
