Problem: https://www.hackerrank.com/challenges/compare-the-triplets/problem?isFullScreen=true

```python
def compareTriplets(a, b):
    Alice=0
    Bob=0
    for x in range(len(a)):
        if a[x] > b[x]:
            Alice+=1
        elif a[x] < b[x]:
            Bob+=1
    return Alice, Bob
a = [1, 2, 3]
b = [3, 2, 1]   
print(compareTriplets(a, b))
print(type(compareTriplets(a, b)))
            
```
