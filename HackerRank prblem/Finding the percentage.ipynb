```python
if __name__ == '__main__':
    n = int(input())
    student_marks = {}
    for _ in range(n):
        name, *line = input().split()
        scores = list(map(float, line))
        student_marks[name] = scores
    query_name = input()
    find_marks=student_marks.get(query_name)
    sum=0
    for x in find_marks:
        sum +=x
    avg=sum/len(find_marks)
    print(f"{avg:.2f}")
    
```
