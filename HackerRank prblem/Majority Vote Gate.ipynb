```python
YES_count=0
NO_count=0
for x in range(int(input())):
    string=input()
    if string.upper()=="YES":
        YES_count+=1
    elif string.upper()=="N" or string.upper()=="NO":
        NO_count+=1
if YES_count >= NO_count:
    print("ACCEPT") 
else:
    print("REJECT")
```
