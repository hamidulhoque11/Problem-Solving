```python
import math
import os
import random
import re
import sys

#
# Complete the 'miniMaxSum' function below.
#
# The function accepts INTEGER_ARRAY arr as parameter.
#

def miniMaxSum(arr):
    arr.sort()
    
    #max
    maxi=arr.copy()
    maxi.pop(0)
    #min
    mini=arr.copy()
    mini.pop()
    #sum
    sum_max=sum(maxi)
    sum_mini=sum(mini)
    print(sum_mini,sum_max)
    
    
    
if __name__ == '__main__':

    arr = list(map(int, input().rstrip().split()))
    miniMaxSum(arr)
```
