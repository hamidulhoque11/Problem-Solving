```python
m,n=list(map(int,input().split()))
value=list(map(int,input().split()))[:m]
a=set(map(int,input().split()))
b=set(map(int,input().split()))
happiness=0
for x in value:
    if x in a :
        happiness+=1
    elif x in b:
        happiness-=1
print(happiness)
```
