```python
'''
Smart Light
Your lab is prototyping a Smart Light that turns on only when the
room is bright enough. You are given two real numbers: the current 
brightness and a threshold. If brightness ≥ threshold, print ON;
otherwise print OFF.'''
n = input()
num = n.split()

brightness = float(num[0])
threshold = float(num[1])

if brightness >= threshold:
    print("ON")
else:
    print("OFF")
```
