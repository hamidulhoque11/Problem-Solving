problem : https://www.hackerrank.com/challenges/string-validators/problem?isFullScreen=true

```python

s = "qA2"
#count
isalnum=0
isalpha=0
isdigit=0
islower=0
isupper=0
for x in s:
    if x.isalnum():
        isalnum+=1
    if x.isalpha():
        isalpha+=1
    if x.isdigit():
        isdigit+=1
    if x.islower():
        islower+=1
    if x.isupper():
        isupper+=1

if isalnum>0:
    print(True)
else:
    print(False)
if isalpha>0:
    print(True)
else:
    print(False)
if isdigit>0:
    print(True)
else:
    print(False)
if islower>0:
    print(True)
else:
    print(False)
if isupper>0:
    print(True)
else:
    print(False)
```
