```python
''' 
link: https://www.hackerrank.com/challenges/python-lists/problem
'''
n=int(input())
lst=[]
for x in range(n):
    command=input().split()
    if command[0]=="insert": #compare with index 0
        lst.insert(int(command[1]),int(command[2]))#list e insert holo
    elif command[0]=="print":
        print(lst)
    elif command[0]=="remove":
        lst.remove(int(command[1]))
    elif command[0]=="append":
        lst.append(int(command[1]))
    elif command[0]=="sort":
        lst.sort()
    elif command[0]=="pop":
        lst.pop()
    elif command[0]=="reverse":
        lst.reverse()
```
