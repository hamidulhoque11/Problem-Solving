```python
def swap_case(s):
    modified=""
    for x in s:
        if x.isupper():
            modified +=x.lower()
        elif x.islower():
            modified+=x.upper()
        else:
            modified+=x
    return modified

if __name__ == '__main__':
    s = input()
    result = swap_case(s)
    print(result)
```
