Add Two Numbers: https://leetcode.com/problems/add-two-numbers/description/

```python
#runtime error :(
class Solution(object):
    def addTwoNumbers(self, l1, l2):
          lOne=0
          lTwo=0
          for x in l1:
               lTwo=int(str(lTwo)+str(x))
          for x in l2:
               lTwo=int(str(lTwo)+str(x))
          sum=lOne+lTwo
          lst=[]
          while 0<sum:
               div=sum//10
               rem=sum%10
               sum=div
               lst.append(rem)
          return lst

l1=map(int,input().split())
l2=map(int,input().split())
value=addTwoNumbers(l1,l2)
value.reverse()
print(value)

```
