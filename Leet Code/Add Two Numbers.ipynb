```python
import numpy as np
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def addTwoNumbers(self, l1: ListNode | None, l2: ListNode | None) -> ListNode | None:
        
        #for l1
        s1=''
        x=0
        while x<=len(l1)-1:
            s1=str(s1)+str(l1[x])
            x+=1
        #for l2
        y=0
        s2=''
        while y<=len(l2)-1:
            s2=str(s2)+str(l2[y])
            y+=1
        #reversed s1 and s2
        rev_s1=s1[::-1]
        rev_s2=s2[::-1]

        # sum s1 and s2
        addi=str(int(rev_s1)+int(rev_s2))

        #reverse sum
        rev_addi=addi[::-1]

        #str to int list
        lst=[]
        x=0
        while x<=len(rev_addi)-1:
            lst.append(int(rev_addi[x]))
            x+=1
        return lst
            
sol=Solution()
l1=[2,4,3]
l2=[5,6,4]

print(sol.addTwoNumbers(l1=l1,l2=l2))
```

```python

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def create_linked_list(lst):
    dummy = ListNode(0)
    current = dummy

    for x in lst:
        current.next = ListNode(x)
        current = current.next

    return dummy.next


class Solution:
    def addTwoNumbers(self, l1, l2):

        # for l1
        s1 = ''
        current = l1

        while current:
            s1 += str(current.val)
            current = current.next

        # for l2
        s2 = ''
        current = l2

        while current:
            s2 += str(current.val)
            current = current.next

        # reverse
        rev_s1 = s1[::-1]
        rev_s2 = s2[::-1]

        # addition
        addi = str(int(rev_s1) + int(rev_s2))

        # reverse sum
        rev_addi = addi[::-1]

        # convert directly to Linked List
        dummy = ListNode(0)
        current = dummy

        x = 0
        while x < len(rev_addi):
            current.next = ListNode(int(rev_addi[x]))
            current = current.next
            x += 1

        return dummy.next


sol = Solution()

l1 = create_linked_list([2,4,3])
l2 = create_linked_list([5,6,4])

result = sol.addTwoNumbers(l1, l2)

while result:
    print(result.val, end=" ")
    result = result.next
```
