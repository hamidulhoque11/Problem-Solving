## Longest Palindromic Substring

https://leetcode.com/problems/longest-palindromic-substring/

```python

s = "abbcccba"

unique = ''
same = ''
x = 0

while x < len(s):

    unique += s[x]

    if unique == unique[::-1]:
        if len(unique) > len(same):
            same = unique

    else:
        # নতুন character যোগ করার পর palindrome না হলে
        # পরের character check করব
        pass

    x += 1

print(same)
```
