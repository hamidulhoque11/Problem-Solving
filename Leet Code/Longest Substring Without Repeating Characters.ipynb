## Longest Substring Without Repeating Characters

https://leetcode.com/problems/longest-substring-without-repeating-characters/description/

```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        unique=''
        unique2=''
        for ch in s:
            if ch not in unique:
                unique+=ch
            else:
                if len(unique)>len(unique2):
                    unique2=unique
                position=unique2.index(ch)
                unique=unique[position+1:]+ch
                
        if len(unique) > len(unique2):
            unique2=unique
        length=len(unique2)
        return len(unique2)
a=Solution()
a.lengthOfLongestSubstring('1R1T7')
```
