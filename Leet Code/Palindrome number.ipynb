Palindrome number: https://leetcode.com/problems/palindrome-number/

```python
class Solution(object):
    def isPalindrome(self, x):
        riv=0
        while 0<x:
            rim=x%10
            div=x//10
            riv=int(str(riv)+str(rim))
            x=div
        if riv==x:
            return True
        else:
            return False

        
```
