```python
class Solution:
    def twoSum(self,nums: list[int], target: int) -> list[int]:
        for x in range(len(nums)):
            for y in range(len(nums)):
                if x==y:
                    continue
                else:
                    if nums[x]+nums[y]==target:
                        result=list(map(int,[x,y])) 
                        return result
    lst=list(map(int,input().split()))
    t=int(input())
    print(twoSum(1,lst,t))
```
