```python
n=int(input())
for x in range(n):
    string=input()
    count=0
    for i in range(len(string)):
        if string[i]=='a' or string[i]=='e' or string[i]=='i' or string[i]=='o' or string[i]=='u' :
            count+=1
            if count>2:
                break
        else:
            count=0
    if count >2:
        print("Happy")
    else:
        print("Sad")
    
```
